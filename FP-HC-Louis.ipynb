Brief Overview: 

The dataset contains information about hate crimes reported across different states in the United States. It provides data that can be used to examine the frequency of hate crimes, compare differences between states, and explore possible relationships between hate crime rates and social, economic, and demographic factors. The dataset includes hate crime statistics reported by the FBI and the Southern Poverty Law Center (SPLC), along with state-level information such as median household income, racial demographics, and political characteristics. Overall, this dataset helps researchers identify patterns and better understand the factors that may be associated with hate crimes across the country.

## Data Loading and Display:

```python
hatecrimes_df = pd.read_csv('hate_crimes.csv')
```

```python
partisan_lean = pd.read_csv('partisan_lean.csv')
```

```python
hatecrimes_df.info()
```

```python
hatecrimes_df = pd.merge(hatecrimes_df, partisan_lean, on = 'state')
```

```python
hatecrimes_clean = hatecrimes_df.rename(columns={'median_household_income': 'med_inc',
                                'share_unemployed_seasonal': 'unemployed',
                                'share_population_in_metro_areas': 'metro_pop',
                                'share_population_with_high_school_degree': 'hs_degree',
                                'share_non_citizen': 'non_citizen',
                                'share_white_poverty': 'white_pov',
                                'gini_index': 'gini',
                                'share_non_white': 'non_white',
                                'share_voters_voted_trump': 'trump_voters',
                                'hate_crimes_per_100k_splc': 'hc_per_100k',
                                'avg_hatecrimes_per_100k_fbi': 'avg_hc_annual',
                                '2022': 'partisan_lean_22'})
```
